# Dataset Analysis & Preparation
Corrected raw dataset locations and reproducible preprocessing.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
ROOT=Path('..'); RAW=ROOT/'03_Data/raw'; PROC=ROOT/'03_Data/processed'; FIG=ROOT/'05_Results/figures'
PROC.mkdir(parents=True,exist_ok=True); FIG.mkdir(parents=True,exist_ok=True)
```

```python
afri=pd.read_csv(RAW/'afrisenti_twi/labeledl_twi_tweets.csv',encoding='cp1252'); afri.columns=['text','label']; afri.text=afri.text.astype(str).str.strip(); afri.label=afri.label.astype(str).str.strip().str.lower(); afri=afri[afri.label.isin(['positive','negative','neutral']) & (afri.text.str.len()>0)].reset_index(drop=True); afri.to_csv(PROC/'afrisenti_twi.csv',index=False); print(afri.shape); print(afri.label.value_counts())
```

```python
parts=[]
for name in ['train.tsv','dev.tsv','test.tsv']:
 d=pd.read_csv(RAW/'asantetwi'/name,sep='\t',encoding='utf-8'); d.columns=['text','label']; parts.append(d)
asante=pd.concat(parts,ignore_index=True); asante.text=asante.text.astype(str).str.strip(); asante.label=asante.label.astype(str).str.strip().str.lower(); asante=asante[asante.label.isin(['positive','negative','neutral']) & (asante.text.str.len()>0)].reset_index(drop=True); asante.to_csv(PROC/'asantetwi_senti.csv',index=False); print(asante.shape); print(asante.label.value_counts())
```

```python
for name,d in [('AfriSenti-Twi',afri),('AsanteTwiSenti',asante)]:
 print(name, 'missing=',d.isna().sum().sum(), 'duplicate_text_rows=',d.text.duplicated().sum(), 'duplicate_full_rows=',d.duplicated().sum())
```
